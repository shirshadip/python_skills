`````python
x,y= sp.symbols('x y')
`````

Algebric manupulation

`````python
sp.expand((x+1)**2)
`````

`````python
sp.factor(x**2+2*x+1)
`````

## Solving equations

`````python
sp.solve(x**2+4,x)
`````

# system of equations

`````python
sp.solve([x+y-2,x-y-0],[x,y])
`````

# Substitution

`````python
expr = x**2+2*x
expr.subs(x,2)
`````

# Calculus

`````python
sp.diff(x**3,x)
`````

## integration

`````python
print(sp.integrate(x**2,x))

function=x**3
#sp.integrate(function, (variable, lower_limit, upper_limit))
sp.integrate(function,(x,0,8))
`````

````Example
Full workflow

`````python
x = sp.symbols('x')
expr= (x+1)**2

expanded= sp.expand(expr)
factored=sp.factor(expanded)
derivate=sp.diff(expr,x)
solution=sp.solve(expr,x)

print ("Expanded",expanded)
print ("Factored:",factored)
print ("Derivative",derivate)
print ("Solution",solution)
`````
