Define integration

```python
from scipy import integrate 
```

```python
def f (x):
    return x **2
result , error = integrate.quad(f,0,2)

print (result)
print (error)
```

infinite limit of integration

```python
import numpy as np 
```

```python
def f(x):
    return 1/x**2
result, error = integrate.quad(f,1,np.inf)
print (result)
print (error)
```

Multivariable integration

```python
def f(x,y):
    return x*y

result, error = integrate.dblquad(
    f,
    0,1, # x limits 
    lambda x:0, #y lower
    lambda x:1 #y upper
)
print (result)
print (error)
```
